# Stratified K-Fold — Keeping the Class Balance in Every Fold

Cross-validation splits the data into folds at random. On a balanced problem that is fine. On an
imbalanced one it is a quiet hazard: a fold can end up with far fewer minority examples than its
share, and a score computed on a handful of positives is mostly noise. In the worst case a fold
contains **none** of the minority class, and the metric is undefined.

**Stratified** K-fold fixes this by construction — each fold is built to preserve the overall class
proportions. It is the default choice for every classification problem, and the cost of using it is
nothing.

## Learning objectives

- Explain why plain `KFold` is risky on imbalanced classification data
- Use `StratifiedKFold` so every fold preserves the class ratio
- Score several model families with `cross_val_score` on a single stratified scheme
- Read the spread of fold scores, not just their mean

### 4.3 Stratified cross-validation

`StratifiedKFold` partitions the data so that **every fold contains the same class proportions as
the full dataset**. If the data is 63/37, each of the 5 folds is 63/37.

Why this matters follows from counting. With plain `KFold` and 5 folds, a fold holds ~114 samples;
the number of minority cases in it is essentially a random draw, and it varies. On this dataset the
consequence is a noisier estimate. On a 99/1 problem the consequence is folds with zero positives,
where precision, recall, and $F_1$ are undefined and quietly return 0 — making a model look broken
when the *split* was broken.

`shuffle=True` randomizes assignment before stratifying, which matters whenever rows arrive sorted
by class or by date.

Two details in the scoring call. `scoring='f1'` reports $F_1$ **on the positive class only**, not
accuracy — the metric that notices when the minority class is being ignored. And rather than the
mean alone, `describe()` shows the full spread across folds; a large `std` says the estimate depends
heavily on which fold you look at, which is exactly the instability stratification is there to
reduce.

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
import numpy as np

# Initialize Stratified KFold
kf = StratifiedKFold(n_splits=5, shuffle=True)

# Evaluate each model using Stratified KFold cross-validation
results = []

for model in models:
    cv_scores = cross_val_score(model, X, y, cv=kf, scoring='f1')
    cv_scores = pd.DataFrame(cv_scores)
    
    print(model)
    display(cv_scores.describe().T)
    print()
# end
```

## 5. Review

- **Random folds do not preserve class balance.** On imbalanced data the minority count per fold is
  a random draw, which adds noise and, in the extreme, produces folds with no positives at all.
- **`StratifiedKFold` builds each fold to match the overall proportions**, at no cost. Use it by
  default for classification.
- **`shuffle=True` matters** when rows arrive in a meaningful order.
- **Score on $F_1$, not accuracy**, when the classes are unbalanced.
- **Read the spread across folds, not only the mean.** A large standard deviation is itself a
  finding about how stable the estimate is.
