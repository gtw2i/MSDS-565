# Real Estate — Exploratory Data Analysis

The opening move of the regression walkthrough. Before any model is fit, the data has to be
understood: what the columns mean, where the values are missing, which fields are miscoded, and
which features look related to the thing we want to predict. Every decision made in
`U1_RealEstate-2_Preprocess` — what to drop, what to fill, what to engineer — is justified by
something found here.

This is a long notebook because EDA is mostly breadth. It works through the numeric columns, then
the categorical ones, then two columns in depth (`Type` and `Nbhd`), then the target itself.

## Learning objectives

- Survey a raw dataset's shape, dtypes, and missingness
- Compare strategies for missing values: drop rows, drop columns, or impute by group
- Read histograms, scatter plots, and a correlation heatmap to find candidate predictors
- Find and repair miscoded categorical levels
- Aggregate with `groupby`, `pivot_table`, and `crosstab` to compare subgroups
- Justify each preprocessing decision with a specific finding

### 4.6 Using .apply() to create a 'Floors' column

```python
def get_floors(x):
    if type(x) == str:
        if '1-STORY' in x:
            return 1
        elif '1-1/2 STORY' in x:
            return 1.5
        elif '2-STORY' in x:
            return 2
        elif '2-1/2 STORY' in x:
            return 2.5
        # end
    # end
    
    return np.nan
# end

df['Floors'] = df.Type.apply(get_floors)
```

### 5.8 Which Neighborhood has the highest ratio of 2-story to 1-story homes?

```python
ct = pd.crosstab( df.Nbhd, df.Type )

one_story = ct.iloc[ :, ct.columns.str.contains("1-STORY") ].sum(axis=1)
two_story = ct.iloc[ :, ct.columns.str.contains("2-STORY") ].sum(axis=1)

(two_story / one_story).sort_values(ascending=False)
```

### 7.1 Fill nans in Lot Frontage column

```python
df['Lot Frontage'].fillna( df['Lot Frontage'].median() ).hist(bins=50)
```

```python
gb = df.groupby(["Nbhd"])['Lot Frontage'].median().sort_values()
gb
```

```python
def fill(r):
    L = r['Lot Frontage']
    
    if pd.isna(L):
        n = r.Nbhd
        return gb[n]
    else:
        return L
    # end
# end

df.apply(fill, axis=1).hist(bins=50)
```

## Review

- **EDA is what justifies preprocessing.** Every drop, fill, and engineered feature in
  `U1_RealEstate-2_Preprocess` traces back to something found in this notebook.
- **Missingness has more than one remedy.** Dropping rows with any NaN, dropping columns with any
  NaN, and imputing each give very different datasets — the counts printed above show how
  different. `Lot Frontage` is filled by **neighborhood median** rather than a global one, because
  the grouping carries real information.
- **Categorical levels are often miscoded.** `1 STORY PUD` versus `1-STORY PUD`, and
  `Bloomington Hts` versus `Bloomington Heights`, are the same category spelled two ways; left
  alone they become separate dummy columns.
- **`groupby`, `pivot_table`, and `crosstab` answer subgroup questions** — median price by
  neighborhood, the neighborhood/type cross-tabulation, and the ratio of 2-story to 1-story homes.
- **The target is right-skewed**, which matters for a squared-error loss: the expensive homes will
  dominate MSE, an argument revisited in `U1_RealEstate-3_Regression`.
- Next: `U1_RealEstate-2_Preprocess` acts on all of this and writes the modeling-ready CSV.
