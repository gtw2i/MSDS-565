# Regular Expressions - Introduction

A regular expression is a small pattern language for finding structure inside raw text. Almost every scraping or parsing job in this unit ends up needing one, because text pulled off the web or out of an instrument file is never already arranged in columns. This notebook works through the syntax on short inline strings, so nothing can fail except the pattern itself.

## Learning objectives

- Match literal characters, digits, and word characters with `re.findall`.
- Control repetition with quantifiers and counted repetition.
- Capture part of a match with a group and pull it out on its own.
- Build patterns for real formats such as phone numbers and dollar amounts.

**Dataset:** none - every example runs on inline strings defined in the notebook.

**References:** https://docs.python.org/3/library/re.html and https://www.w3schools.com/python/python_regex.asp

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

pd.set_option('display.max_columns',100)
pd.set_option('display.max_rows',100)

plt.style.use('dark_background')

import warnings
warnings.filterwarnings('ignore')

# Shared course helpers (msds565_helpers.py lives in the repo root).
# Notebooks sit two folders below the root, so '../..' points back to it.
import sys
sys.path.append('../..')
import msds565_helpers as helpers

import re
```

```python
text = "If you know Python,    then\nyou're power level is over 9000"
print(text)
```

```python
# find all o's
re.findall(r"o", text)
```

```python
# find all ou's
re.findall(r"ou", text)
```

```python
# find all numbers
re.findall(r"\d+", text)
```

```python
# find all 2-digit numbers numbers
re.findall(r"\d{2}", text)
```

```python
# find all word characters
re.findall(r"\w", text)
```

```python
# find all words (no contractions)
re.findall(r"\w+", text)
```

```python
# find all non-space characters
re.findall(r"\S+", text)
```

```python
# find all words (with contractions)
re.findall(r"[\w']+", text)
```

```python
# capture only digits using parentheses
text = "If you know Python,    then\nyou're 4654 power level is over 9000. over 60"

re.findall(r"over (\d+)", text)
```

```python
# find phone numbers
text = "my number is (615) 555-1234. her number is (012) 345-6789"

re.findall(r"\(\d{3}\) \d{3}-\d{4}", text)
```

```python
# find prices
text = "I made $12.21 she made $4,001.99 he made $1,033,080.01"

matches = re.findall(r"\$([\d,]+\.\d{2})", text)
matches = [ re.sub(',', '', m) for m in matches ]
matches = [ float(m[1:]) for m in matches ]
matches
```

## Review

Regex is a language for describing the *shape* of text rather than its content, and these are the pieces it is built from.

| Piece | Matches |
|---|---|
| `\d`, `\w`, `\s` | a digit, a word character, whitespace |
| `+`, `*`, `?` | one-or-more, zero-or-more, optional |
| `{n}`, `{n,m}` | exactly n, or between n and m |
| `[...]` | any one character from the set |
| `(...)` | a capture group - the part you want handed back |

**Takeaways**

- **`findall` returns your groups when you use them, and the whole match when you don't.** That switch is the most common source of surprise for people new to `re`.
- **Build a pattern incrementally.** Match the easy part, print what came back, then tighten it. Writing the whole expression before testing any of it is how regex earns its reputation.
- **A pattern that is hard to read now will be impossible to fix later.** When one gets long, doing the job in two simple passes is usually the better engineering.
