# Regular Expressions - Extracting Links from HTML

This notebook pulls every hyperlink out of a saved HTML page twice: once by hand, slicing strings between quote characters, and once with a single regular expression. Both give the same answer, and the contrast is the whole lesson - regex turns fifteen lines of fragile index arithmetic into one line you can actually read.

## Learning objectives

- Read a saved HTML file into Python as raw text.
- Extract `href` values by manual string slicing, and explain where that approach breaks.
- Do the same extraction with one `re.findall` pattern.
- Judge when a regex is the right tool for HTML and when a parser is.

## Background

This is the same `links.html` file that `U3-3_Scrape-1_Intro.ipynb` parses with BeautifulSoup, so the two notebooks are worth reading together. Regex is fast and blunt; a parser understands the document's structure. For pulling one attribute out of flat markup, regex is fine - for navigating a real page, it is not.

## 1. Without regex

```python
lines = text.split('\n')
lines
```

```python
hrefLines = []
for line in lines:
    if 'href' in line:
        hrefLines.append(line.strip())
    # end
# end
hrefLines
```

```python
for line in hrefLines:
    idx1 = line.find('\"')+1
    idx2 = line.find('\"', idx1)
    
    print(line[idx1:idx2])
```

## 2. With regex

```python
re.findall(r"href=\"(.*)\"", text)
```

## 3. Review

Two routes to the same list of links, and a clear verdict on when each one is appropriate.

**Takeaways**

- **The manual version hides its assumptions** - that quotes are double, that `href` comes first in the tag, that nothing else on the line contains a quote. The regex makes the same assumptions, but visibly and in one place.
- **Regex is the right tool for a flat, predictable extraction** from text you control or have already saved to disk.
- **A parser is the right tool for a real page.** Nested tags, optional attributes, and malformed markup are precisely what regex cannot handle and what BeautifulSoup exists for - see `U3-3_Scrape-1_Intro.ipynb`.
