# LLM LangChain — Unified Abstraction Layer

**LangChain** wraps every provider behind one interface: `llm.invoke(prompt).content`. Compare this single line to the provider-specific calls in `U3-2_LLM-1_Intro.ipynb` through `U3-2_LLM-4_LangchainRAG.ipynb`.

Install packages if needed: `pip install langchain-openai langchain-anthropic langchain-google-genai langchain-ollama`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

pd.set_option('display.max_columns',100)
pd.set_option('display.max_rows',100)

plt.style.use('dark_background')

import warnings
warnings.filterwarnings('ignore')

# Shared course helpers (msds565_helpers.py lives in the repo root).
# Notebooks sit two folders below the root, so '../..' points back to it.
import sys
sys.path.append('../..')
import msds565_helpers as helpers

import os

PROMPT = 'Create a random number.'
```

---
## Ollama (local)

```python
from langchain_ollama import ChatOllama

OLLAMA_MODEL = 'llama3.2:3b-instruct-q5_K_M'

ollama_llm = ChatOllama(model=OLLAMA_MODEL)
print(ollama_llm.invoke(PROMPT).content)
```

---
## OpenAI

```python
from langchain_openai import ChatOpenAI

OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
OPENAI_MODEL   = 'gpt-5'

openai_llm = ChatOpenAI(model=OPENAI_MODEL, api_key=OPENAI_API_KEY)
print(openai_llm.invoke(PROMPT).content)
```

---
## Anthropic

```python
from langchain_anthropic import ChatAnthropic

ANTHROPIC_API_KEY = os.getenv('ANTHROPIC_API_KEY')
ANTHROPIC_MODEL   = 'claude-sonnet-4-5-20250929'

anthropic_llm = ChatAnthropic(model=ANTHROPIC_MODEL, api_key=ANTHROPIC_API_KEY)
print(anthropic_llm.invoke(PROMPT).content)
```

---
## Google Gemini

```python
from langchain_google_genai import ChatGoogleGenerativeAI

GEMINI_API_KEY = os.getenv('GEMINI_API_KEY')
GOOGLE_MODEL   = 'gemini-2.5-flash'

gemini_llm = ChatGoogleGenerativeAI(model=GOOGLE_MODEL, google_api_key=GEMINI_API_KEY)
print(gemini_llm.invoke(PROMPT).content)
```

---
## All providers — one loop

Because every provider shares the `.invoke()` interface, looping over all four is trivial — something you couldn't do cleanly with native SDKs.

```python
llms = {
    'Ollama':    ollama_llm,
    'OpenAI':    openai_llm,
    'Anthropic': anthropic_llm,
    'Gemini':    gemini_llm,
}
for name, llm in llms.items():
    print(f'── {name} ──')
    print(llm.invoke(PROMPT).content)
    print()
```

## Review

**Takeaways**

- **One interface, four providers.** `llm.invoke(prompt).content` is identical no matter which model is behind it, so swapping providers becomes a one-line change instead of a rewrite.
- **The abstraction costs you the provider's newest features.** Anything not yet wrapped is reachable only through the native SDK, which is why the earlier notebooks in this section still matter.
- **Use LangChain when you want composition** - chains, retrievers, memory, and the RAG pipeline built in `U3-2_LLM-4_LangchainRAG.ipynb`. For a single call, the native SDK is simpler and has one less dependency.
- **A uniform interface makes fair comparison easy**, which is what the final loop over all four providers demonstrates.
