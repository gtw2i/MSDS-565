# Scraping Wikipedia Tables — Building a Dataset from Linked Pages

Most useful scrapes are two levels deep: one page lists the things you care about, and each of those things has its own page holding the actual fields. Here the list is Wikipedia's *List of computer scientists*, and the fields are whatever each person's infobox happens to carry. The result is a real DataFrame — sparse, inconsistent, and needing exactly the kind of cleaning that scraped data always needs.

## Learning objectives

- Extract an entity list and its links from a Wikipedia list page.
- Follow each link and parse the target page's infobox into a record.
- Assemble heterogeneous records into a DataFrame and clean its columns.
- Explain why a two-level scrape produces sparse data, and how to decide which columns to keep.

```python
base_url = 'https://en.wikipedia.org/'
new_url = 'wiki/List_of_computer_scientists'
url = base_url + new_url

response = requests.get(url)

html = response.text
soup = BeautifulSoup(html, 'html.parser')
print(soup.prettify()[:1000])
```

## 1. Get the names and links

```python
links = {}

div = soup.find('div', class_="mw-content-ltr mw-parser-output")
for ul in div.find_all('ul')[:-3]:
    try:
        for li in ul.find_all("li")[:5]:
            a = li.find('a')
            href = a['href']
            links[a.text] = href
        # end
    except:
        1
    # end
# end
```

## 2. Scrape each person's page

```python
count = 0
cs_data = {}
for k, v in tqdm(links.items(), ncols=50):
    #print(k)
    new_url = v
    url = base_url + new_url
    response = requests.get(url)
    html = response.text
    soup = BeautifulSoup(html, 'html.parser')

    info = {}
    data = soup.find('table', class_="infobox biography vcard")
    try:
        for tr in data.find_all('tr'):
            try:
                a = tr.find('th').text
                
                #b = tr.find('td').text

                b = []
                sp = tr.find('td').find_all("span")
                if len(sp)>0:
                    b += [ item.text for item in sp ]
                    #b = ",".join(b)
                
                #"""
                li = tr.find('td').find_all("li")
                if len(li)>0:
                    b += [ item.text for item in li ]
                    #b = ",".join(b)
                
                href = tr.find('td').find_all("a")
                if len(href)>0:
                    b += [ item.text for item in href ]
                    #b = ",".join(b)
                #""";
                
                b = ",".join(b)
                
                #print(len(sp),len(li),len(href))
                info[a] = b
            except:
                1
            # end
        # end
        count += 1
    except:
        1
    # end
    cs_data[k] = info

    #if count >= 20:
        #break
# end
```

## 3. Clean the columns

```python
df = pd.DataFrame(cs_data).T

df.info()
df.head()
```

```python
df.columns = pd.Series(df.columns).apply(
    lambda x: x.replace('\xa0', ' ').strip() if type(x)==str else x
)
```

```python
for c in df.columns:
    df[c] = df[c].apply(
        lambda x: x.replace('\xa0', ' ').strip() if type(x)==str else x
    )
# end

for c in df.columns[1:]:
    df[c] = df[c].apply(
        lambda x: re.sub(r"\[[\d,]+\]", "", x) if type(x)==str else x
    )
# end

df.head()
```

```python
df = df.dropna(thresh=df.shape[0]//4, axis=1)
df = df.dropna(thresh=df.shape[1]//4, axis=0)
df.head()
```

```python
def fix_born(x):
    if type(x)==str:
        y = re.findall(r"\d{4}-\d{2}-\d{2}", x)
        if len(y)>0:
            return y[0]
    # end
    return np.nan
# end

df.Born = df.Born.apply( fix_born )
```

```python
df.Fields = df.Fields.str.lower()
df['Known for'] = df['Known for'].str.lower()
```

## 4. Review

A two-level scrape: one list page gives you the entities, and each entity's own page gives you the fields.

**Takeaways**

- **The list page is an index, not the data.** Its value is the set of links it hands you; everything worth having is one hop further in.
- **Infobox parsing is where real datasets come from and where they get messy** — different people's pages carry different fields, so the assembled frame is sparse and needs the column cleaning done here.
- **Drop columns by how well populated they are**, not by guessing which ones matter; a field present on four of six hundred pages is not a feature.
- **Be polite and be patient.** Several hundred sequential requests is the point at which rate limiting, timeouts, and per-item error handling stop being optional.
