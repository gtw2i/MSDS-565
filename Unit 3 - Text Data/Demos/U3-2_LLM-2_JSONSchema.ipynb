# LLM JSON Schema — Structured Output

An LLM that answers in prose is useless to the rest of your program. This notebook extracts structured metadata from a real academic paper PDF, constraining each provider's output to the same JSON schema so the response always parses cleanly into a Python dict — no regex, no hallucinated keys, no defensive string surgery.

## Learning objectives

- Describe an expected response with a JSON Schema.
- Constrain model output to that schema on four different provider SDKs.
- Extract text from a PDF and pass it to a model as context.
- Explain why schema-constrained output is what makes an LLM callable from ordinary code.

## Background

The four provider SDKs and their environment-variable keys are the ones set up in `U3-2_LLM-1_Intro.ipynb`. Each provider has its own mechanism for enforcing a schema; all four share the same schema and the same prompt here, so the differences you see are purely in the plumbing.

**Prerequisites:** `U3-2_LLM-1_Intro.ipynb`

**Dataset:** `Towards_a_Platform_for_AI_Assisted_Papyrology.pdf` and `paper_schema.json`, both sitting beside this notebook.

**Requirements:** `OPENAI_API_KEY`, `ANTHROPIC_API_KEY`, and `GEMINI_API_KEY` in the environment for the cloud sections, and a local Ollama server with `llama3.2:3b-instruct-q5_K_M` for the last one. Run only the sections you have access to — they are independent.

**References:** https://json-schema.org/

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

pd.set_option('display.max_columns',100)
pd.set_option('display.max_rows',100)

plt.style.use('dark_background')

import warnings
warnings.filterwarnings('ignore')

# Shared course helpers (msds565_helpers.py lives in the repo root).
# Notebooks sit two folders below the root, so '../..' points back to it.
import sys
sys.path.append('../..')
import msds565_helpers as helpers

import json
import os
from pypdf import PdfReader

with open('paper_schema.json') as f:
    SCHEMA = json.load(f)

reader = PdfReader('Towards_a_Platform_for_AI_Assisted_Papyrology.pdf')
pdf_text = '\n'.join(page.extract_text() for page in reader.pages)

SCHEMA_PROMPT = f"""
Extract structured metadata from the following academic paper:

{pdf_text}
"""
```

---
## 1. OpenAI

Passed as `response_format` with `type='json_schema'`. The SDK validates the response automatically when `strict=True`.

```python
from openai import OpenAI

OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
OPENAI_MODEL   = 'gpt-5'

openai_client = OpenAI(api_key=OPENAI_API_KEY)
openai_resp = openai_client.chat.completions.create(
    model=OPENAI_MODEL,
    messages=[
        {
            'role': 'user',
            'content': SCHEMA_PROMPT
        }
    ],
    response_format={
        'type': 'json_schema',
        'json_schema': {
            'name': 'book_recommendation',
            'strict': True,
            'schema': SCHEMA
        }
    }
)
result = json.loads(
    openai_resp.choices[0].message.content
)
print(json.dumps(result, indent=2))
```

---
## 2. Anthropic

Uses `output_config` to attach the JSON schema directly to the request. The SDK also offers `client.messages.parse()` for automatic Pydantic model validation.

```python
from anthropic import Anthropic

ANTHROPIC_API_KEY = os.getenv('ANTHROPIC_API_KEY')
ANTHROPIC_MODEL   = 'claude-sonnet-4-5-20250929'

anthropic_client = Anthropic(api_key=ANTHROPIC_API_KEY)
anthropic_resp = anthropic_client.messages.create(
    model=ANTHROPIC_MODEL,
    max_tokens=500,
    messages=[{'role': 'user', 'content': SCHEMA_PROMPT}],
    output_config={'format': {'type': 'json_schema', 'schema': SCHEMA}}
)
result = json.loads(anthropic_resp.content[0].text)
print(json.dumps(result, indent=2))
```

---
## 3. Google Gemini

Pass the schema to `response_schema` in `GenerateContentConfig` (alongside `response_mime_type='application/json'`). Gemini enforces the structure natively — no need to inject the schema into the prompt. (`additionalProperties` is dropped, since Gemini's schema subset doesn't accept it.)

```python
from google import genai
from google.genai import types

GEMINI_API_KEY = os.getenv('GEMINI_API_KEY')
GOOGLE_MODEL   = 'gemini-2.5-flash'

client = genai.Client(api_key=GEMINI_API_KEY)

# Gemini's response_schema follows the OpenAPI subset and rejects `additionalProperties`
# (kept in SCHEMA for OpenAI strict mode), so drop that one key.
gemini_schema = {k: v for k, v in SCHEMA.items() if k != 'additionalProperties'}

gemini_resp = client.models.generate_content(
    model=GOOGLE_MODEL,
    contents=SCHEMA_PROMPT,
    config=types.GenerateContentConfig(
        response_mime_type='application/json',
        response_schema=gemini_schema,
    ),
)
result = json.loads(gemini_resp.text)
print(json.dumps(result, indent=2))
```

---
## 4. Ollama (local)

Pass the JSON schema directly to the `format=` parameter. Ollama constrains the tokenizer at the grammar level — the output is always valid JSON.

```python
import ollama

OLLAMA_MODEL = 'llama3.2:3b-instruct-q5_K_M'

ollama_resp = ollama.chat(
    model=OLLAMA_MODEL,
    messages=[{'role': 'user', 'content': SCHEMA_PROMPT}],
    format=SCHEMA,
    options={'temperature': 0.2}
)
result = json.loads(ollama_resp['message']['content'])
print(json.dumps(result, indent=2))
```

## 5. Review

The same paper, the same schema, and the same prompt, sent four different ways — the differences are entirely in how each provider is told to constrain its output.

| Provider | How the schema is supplied |
|---|---|
| OpenAI | A JSON-schema response format on the request |
| Anthropic | The schema described in the prompt, with the model asked to emit only JSON |
| Google Gemini | A response schema and MIME type on the generation config |
| Ollama | A `format` argument carrying the schema |

**Takeaways**

- **A schema turns an LLM into something you can call from code.** Without one you are writing a parser against prose that changes wording every run; with one the response loads straight into a dict.
- **Every provider supports this, and every provider spells it differently.** The schema is portable; the plumbing around it is not.
- **The schema is documentation as well as constraint** — `paper_schema.json` states exactly what a caller is entitled to expect back.
- **Validate anyway.** Constrained decoding makes malformed output rare, not impossible, and a missing key found at parse time is much cheaper than one found three functions later.
