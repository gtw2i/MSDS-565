## Background

An HTML document is a tree of **tags**: `<html>` contains `<head>` and `<body>`, a bulleted list `<ul>` contains list items `<li>`, and so on. Tags carry **attributes** inside the angle brackets — in `<a href="https://www.google.com">Google</a>`, the anchor tag's `href` attribute holds the link target, and `Google` is its visible text. Scraping is mostly asking for tags by name, then reading off their attributes and text.

## 1. Read and parse the saved page

`links.html` is read from disk like any other text file, and the raw HTML string is handed to `BeautifulSoup`, which parses it into a searchable tree. `prettify()` re-prints the document with indentation that makes the nesting visible: a `<ul>` list of `<li>` items, each wrapping one `<a>` link.

```python
data_folder = 'C:/Users/Graham West/Python Notebooks/Meharry Teaching/Datasets/'

with open(data_folder + "Text Processing/links.html",'r') as f:\
    text = f.read()
# end

soup = BeautifulSoup(text, "html.parser")

print(soup.prettify())
```

## 2. Extract every link

`find_all('a')` returns every anchor tag in the document. A tag can be indexed like a dictionary of its attributes, so `a['href']` reads each link's target, and the loop collects all four into an ordinary Python list.

```python
links = []
for a in soup.find_all("a"):
    links.append(a['href'])
links
```

## 3. Review

This notebook covered the core BeautifulSoup workflow, practiced on a page that cannot change out from under us:

- **Parse:** `BeautifulSoup(text, 'html.parser')` turns raw HTML into a searchable tree; `prettify()` displays its nesting.
- **Search:** `find_all('a')` returns every matching tag in the document.
- **Extract:** indexing a tag like a dictionary (`a['href']`) reads one of its attributes; `.text` would read its visible text.
